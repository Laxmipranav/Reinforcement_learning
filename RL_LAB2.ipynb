```python
import random

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of machines
M = 3

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# Part B – Initialize N, R and Q
N = [0, 0, 0]       # Number of times each machine is selected
R = [0, 0, 0]       # Total reward from each machine
Q = [0.0, 0.0, 0.0] # Estimated Q-values

print("===== Epsilon-Greedy Multi-Armed Bandit =====")

# Part C – Epsilon-Greedy Decision
for play in range(1, N_play + 1):

    # Generate random number between 0 and 1
    r = random.random()

    if r < epsilon:
        # Exploration
        selected = random.randint(0, M - 1)
        decision = "Exploration"
    else:
        # Exploitation
        selected = Q.index(max(Q))
        decision = "Exploitation"

    print("\nPlay", play)
    print("Decision:", decision)
    print("Selected Machine:", machines[selected])

    # Part D – Reward Collection
    reward = float(input("Enter reward: "))

    # Update number of times selected
    N[selected] += 1

    # Update total reward
    R[selected] += reward

    # Part E – Incremental sample-average Q-value
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    print("Updated Q-value:", Q[selected])

# Part F – Display Results
print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tQ")

for i in range(M):
    print(machines[i], "\t", N[i], "\t", R[i], "\t", round(Q[i], 2))

# Total reward
total_reward = sum(R)

print("\nTotal Reward after 10 plays =", total_reward)
```
